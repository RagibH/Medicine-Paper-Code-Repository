```python
import os
import shutil
import pandas as pd

# -------- PATHS --------
CSV_PATH = r"G:\Ragib\Prescription\Cross Testing Data\testing_labels.csv"
TEST_IMAGES_DIR = r"G:\Ragib\Prescription\Cross Testing Data\testing_words"

# Use your dataset (raw or Train split) to get class names
YOUR_DATASET_DIR = r"G:\Ragib\Prescription\Final Dataset For journal work"

OUTPUT_DIR = r"G:\Ragib\Prescription\Cross_Test_Final"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------- STEP 1: LOAD YOUR DATASET CLASSES --------
dataset_classes = [
    cls.strip()
    for cls in os.listdir(YOUR_DATASET_DIR)
    if os.path.isdir(os.path.join(YOUR_DATASET_DIR, cls))
]

# Case-insensitive matching
dataset_class_map = {cls.lower(): cls for cls in dataset_classes}

print(f"✔ Found {len(dataset_class_map)} classes in your dataset")

# -------- STEP 2: LOAD CSV --------
df = pd.read_csv(CSV_PATH)

print("CSV columns:", df.columns.tolist())
# Expected: ['IMAGE', 'MEDICINE_NAME', 'GENERIC_NAME']

# -------- STEP 3: FILTER & COPY --------
matched = 0
skipped = 0

for _, row in df.iterrows():
    img_name = str(row["IMAGE"]).strip()
    med_name = str(row["MEDICINE_NAME"]).strip()

    med_key = med_name.lower()

    # Check if medicine exists in your dataset
    if med_key not in dataset_class_map:
        skipped += 1
        continue

    src_img = os.path.join(TEST_IMAGES_DIR, img_name)
    if not os.path.exists(src_img):
        skipped += 1
        continue

    # Use exact folder name from your dataset
    class_folder = dataset_class_map[med_key]
    dst_dir = os.path.join(OUTPUT_DIR, class_folder)
    os.makedirs(dst_dir, exist_ok=True)

    dst_img = os.path.join(dst_dir, img_name)
    if not os.path.exists(dst_img):
        shutil.copy(src_img, dst_img)
        matched += 1

print("\n✅ Cross-testing dataset created successfully")
print(f"✔ Images copied: {matched}")
print(f"✖ Images skipped: {skipped}")
print(f"📁 Output folder: {OUTPUT_DIR}")
```
